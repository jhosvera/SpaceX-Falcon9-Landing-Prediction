**Objetivo:** explorar qué variables influyen en el éxito del aterrizaje (número de vuelo, carga útil, sitio, órbita y año).
Las figuras se guardan en `figures/`.

```python
BASE = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/"
import os

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid")
os.makedirs("../figures", exist_ok=True)
df = pd.read_csv(BASE + "dataset_part_2.csv")
df.head()
```

## 1 · Número de vuelo vs. sitio de lanzamiento

```python
g = sns.catplot(y="LaunchSite", x="FlightNumber", hue="Class", data=df, aspect=3)
g.set_axis_labels("Número de vuelo", "Sitio de lanzamiento")
g.savefig("../figures/01_flightnumber_vs_launchsite.png", dpi=150)
plt.show()
```

## 2 · Masa de carga útil vs. sitio de lanzamiento

```python
g = sns.catplot(y="LaunchSite", x="PayloadMass", hue="Class", data=df, aspect=3)
g.set_axis_labels("Masa de carga útil (kg)", "Sitio de lanzamiento")
g.savefig("../figures/02_payload_vs_launchsite.png", dpi=150)
plt.show()
```

## 3 · Tasa de éxito por tipo de órbita

```python
success_by_orbit = df.groupby("Orbit")["Class"].mean().sort_values(ascending=False)
ax = success_by_orbit.plot(kind="bar", figsize=(9, 4), color="#D9480F")
ax.set_ylabel("Tasa de éxito"); ax.set_xlabel("Órbita"); ax.set_title("Tasa de éxito por órbita")
plt.tight_layout(); plt.savefig("../figures/03_success_by_orbit.png", dpi=150); plt.show()
success_by_orbit
```

## 4 · Número de vuelo vs. órbita y carga útil vs. órbita

```python
g = sns.catplot(y="Orbit", x="FlightNumber", hue="Class", data=df, aspect=3)
g.savefig("../figures/04_flightnumber_vs_orbit.png", dpi=150)
g = sns.catplot(y="Orbit", x="PayloadMass", hue="Class", data=df, aspect=3)
g.savefig("../figures/05_payload_vs_orbit.png", dpi=150)
plt.show()
```

## 5 · Tendencia anual de la tasa de éxito

```python
df["Year"] = pd.to_datetime(df["Date"]).dt.year
yearly = df.groupby("Year")["Class"].mean()
ax = yearly.plot(marker="o", figsize=(9, 4), color="#0F8B8D")
ax.set_ylabel("Tasa de éxito"); ax.set_xlabel("Año"); ax.set_title("Tendencia anual de éxito")
plt.tight_layout(); plt.savefig("../figures/06_yearly_trend.png", dpi=150); plt.show()
yearly
```

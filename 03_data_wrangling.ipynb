**Objetivo:** crear la variable objetivo `Class` (1 = aterrizaje exitoso, 0 = fallido), revisar sitios y órbitas
y generar la matriz de variables con codificación one-hot (83 columnas).

```python
import numpy as np
import pandas as pd

pd.set_option("display.max_columns", None)
```

```python
BASE = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/"
df = pd.read_csv(BASE + "dataset_part_1.csv")
df.head()
```

## Valores nulos y distribución por sitio, órbita y resultado

```python
print((df.isnull().sum() / len(df) * 100).round(2))
print()
print(df["LaunchSite"].value_counts(), end="\n\n")
print(df["Orbit"].value_counts(), end="\n\n")
landing_outcomes = df["Outcome"].value_counts()
landing_outcomes
```

## Variable objetivo `Class`

```python
bad_outcomes = {"False ASDS", "False Ocean", "False RTLS", "None ASDS", "None None"}
df["Class"] = df["Outcome"].apply(lambda o: 0 if o in bad_outcomes else 1)
print("Tasa de éxito:", round(df["Class"].mean(), 4))
df[["Class"]].head()
```

```python
df.to_csv("../data/dataset_part_2.csv", index=False)
```

## Ingeniería de variables y codificación one-hot

```python
features = df[["FlightNumber", "PayloadMass", "Orbit", "LaunchSite", "Flights", "GridFins",
               "Reused", "Legs", "LandingPad", "Block", "ReusedCount", "Serial"]]

features_one_hot = pd.get_dummies(features, columns=["Orbit", "LaunchSite", "LandingPad", "Serial"])
features_one_hot = features_one_hot.astype("float64")
print("Forma de la matriz de variables:", features_one_hot.shape)
features_one_hot.head()
```

```python
features_one_hot.to_csv("../data/dataset_part_3.csv", index=False)
```
